```python
# Напишем функцию для обработки столбцов с датами
def process_to_datetime(column):
    if '_at' in column.name:
        return pd.to_datetime(column)
    return column
```

```python
# Напишем функцию для обработки в датафрейме
def process_dataset(data, column_processor):
    for column_name in data.columns:  # Используем data, а не dataset
        data[column_name] = column_processor(data[column_name])
    return data
```

### 3.4. Цены стартапов по категориям

Категории стартапов с типично высокими ценами покупки стартапов и значительным разбросом цен могут быть привлекательными для крупных инвесторов, которые готовы к высоким рискам ради потенциально больших доходов. Среди категорий стартапов выделите категории стартапов, характеризующиеся:

- типично высокими ценами;
- и наибольшим разбросом цен за стартап.

Объясните, почему решили составить топ именно из такого числа категорий и почему рассчитывали именно так.

```python
# Объединяем данные
merged = pd.merge(main_data, acquisition, left_on='company_id', right_on='acquired_company_id', how='inner')
```

```python
# Фильтруем только сделки с положительной ценой
merged = merged[merged['price_amount'] > 0]
```

```python
# Ограничиваем данные 90-м процентилем, чтобы исключить сильные выбросы
price_90th = merged['price_amount'].quantile(0.90)
filtered_merged = merged[merged['price_amount'] <= price_90th]
```

```python
# Делаем копию
filtered_merged = filtered_merged.copy()
```

```python
# Группируем по категориям и считаем медиану, стандартное отклонение и количество сделок
category_stats = filtered_merged.groupby('category_code')['price_amount'].agg(['median', 'std', 'count'])
```

```python
# Фильтруем категории с минимум 10 сделками
category_stats = category_stats[category_stats['count'] >= 10]
```

```python
# Определяем топ-5 категорий по медианной цене
top_median = category_stats[['median', 'count']].sort_values('median', ascending=False).head(5)
```

```python
# Определяем топ-5 категорий по разбросу цен
top_std = category_stats[['std', 'count']].sort_values('std', ascending=False).head(5)
```

```python
# Визуализация медианной цены
plt.figure(figsize=(12, 5))
sns.barplot(x=top_median.index, y=top_median['median'], palette='viridis')

# Подписываем оси
plt.title('Топ-5 категорий по медианной цене покупки стартапов')
plt.xlabel('Категория стартапов')
plt.ylabel('Медианная цена')

# Поворачиваем подписи
plt.xticks(rotation=45)

# Добавляем сетку
plt.grid(axis='y')
```

```python
# Визуализация разброса цен
plt.figure(figsize=(12, 5))
sns.barplot(x=top_std.index, y=top_std['std'], palette='magma')

# Подписываем оси
plt.title('Топ-5 категорий с наибольшим разбросом цен')
plt.xlabel('Категория стартапов')
plt.ylabel('Стандартное отклонение цены')

# Поворачиваем подписи
plt.xticks(rotation=45)

# Добавляем сетку
plt.grid(axis='y')
```

```python
# Логарифмируем цену для лучшей читаемости распределения
filtered_merged['log_price'] = np.log1p(filtered_merged['price_amount'])
```

```python
# Построим boxplot распределения цен в топ-5 категориях по медиане
plt.figure(figsize=(12, 6))
sns.boxplot(x='category_code', y='log_price', data=filtered_merged[filtered_merged['category_code'].isin(top_median.index)])

# Подписываем оси
plt.title('Распределение логарифма цен в топ-5 категориях по медиане')
plt.xlabel('Категория стартапов')
plt.ylabel('Логарифм цены')

# Поворачиваем подписи
plt.xticks(rotation=45)

# Добавляем сетку
plt.grid(axis='y')
```

Вывод:

Выделены топ-5 категорий стартапов с высокими медианными ценами и значительным разбросом цен, что делает их привлекательными для крупных инвесторов. Топ-5 по медиане демонстрирует категории с высокими средними ценами, а топ-5 по стандартному отклонению — категории с наибольшим разбросом цен, что указывает на высокие риски и потенциально большие прибыли. Визуализация данных помогает выявить наиболее интересные для инвесторов категории.
